## Introdução

A Compensação Financeira pela Exploração de Recursos Minerais (CFEM) é um contraprestação financeira cobrado das empresas mineradoras como contraprestação pelo uso econômico dos recursos minerais da União. A Agência Nacional de Mineração (ANM) é responsável por normatizar, fiscalizar e operacionalizar a arrecadação e distribuição desses recursos.

O conjunto disponibilizado pela ANM abrange informações de arrecadação, autuação e distribuição da CFEM, atualizadas diariamente e organizadas em arquivos separados por períodos de anos para facilitar o tratamento do grande volume de registros. Esses dados se inserem no contexto de transparência e controle social sobre a mineração, política fiscal e desenvolvimento regional no Brasil.

No cenário atual de instabilidade geopolítica, marcado por conflitos internacionais, tensões comerciais e incertezas econômicas globais, o setor mineral tem despertado crescente interesse por parte de investidores. Recursos minerais estratégicos, como ouro, cobre e ferro, passam a ser vistos como ativos de segurança e proteção de capital, além de insumos essenciais para cadeias produtivas globais. Nesse contexto, a arrecadação da CFEM e os dados divulgados pela ANM tornam-se ainda mais relevantes para a análise de riscos, oportunidades de investimento e planejamento econômico de longo prazo no país.


### Sobre o conjunto de dados:

Os dados são de acesso aberto e estão disponíveis neste [link](https://dados.gov.br/dados/conjuntos-dados/sistema-arrecadacao)
. Neste projeto, foram trabalhados os dados dispostos no arquivo “CFEM_Arrecadacao_2022_2026.csv”, que registram detalhes de cada guia de recolhimento da CFEM, tendo como principais colunas:

- Ano e mês: Período do fato gerador ou de criação da guia de pagamento.


- Número do processo minerário: Identificador único do processo na ANM.
​

- Tipo de titular: Indicador de pessoa física (PF), jurídica (PJ) ou outra categoria do recolhedor.
​

- Substância mineral: Nome da substância explorada (ex.: minério de ferro, ouro, bauxita).
​

- Quantidade comercializada: Volume da substância em unidade de medida específica (ex.: toneladas).
​

- Valor recolhido (em R$): Montante efetivamente pago de CFEM.
​

- Município/UF de extração: Local geográfico da atividade minerária.
​

## Análise-Exploratória

A análise exploratória se inicia logo após a realização de um tratamento 
A figura abaixo, exibe os Estados com maiores arrecadações de receita para a união no ano de 2025. Segundo os dados aberto da CFEM.

![image.png](https://github.com/MarcosVinicio78542/Panorama-da-Arrecada-o-da-CFEM-no-Brasil-2022-2026-/blob/main/Plot1.png?raw=true)

Como resultado temos que em 2025 os Estados de Minas Gerais e Pará dominam boa parte das receitas arrecadados pela mineração no Brasil na casa dos Bilhões de reais. Com Roráima e Acre os dois Estados que apontaram menores receitas neste ano. Já no gráfico a seguir, apontam os quinze recurços minerais com maiores contribuições na receira em 2025.


![image.png](https://github.com/MarcosVinicio78542/Panorama-da-Arrecada-o-da-CFEM-no-Brasil-2022-2026-/blob/main/plot2.png?raw=true)

Como visto acima, o ferro, o minério de ferro, o cobre, o minério de ouro e o ouro constituem grande parcela do valor arrecadado pelas atividades de mineração em 2025, tanto por empresas PJ quanto por PF.
O minério é a rocha ou material natural extraído da natureza que contém o metal de interesse ainda misturado a outros minerais e impurezas. No caso do minério de ouro por exemplo, ele contém partículas de ouro dispersas em sua estrutura. Essas partículas são pequenas porções do metal que só podem ser aproveitadas após processos de extração e beneficiamento.
Já o ouro é o metal propriamente dito, obtido depois da separação e purificação dessas partículas, apresentando alta pureza, brilho característico e elevado valor econômico.

![image.png](https://github.com/MarcosVinicio78542/Panorama-da-Arrecada-o-da-CFEM-no-Brasil-2022-2026-/blob/main/plot3.png?raw=true)

No gráfico acima, observa-se que os dois últimos anos apresentaram elevados níveis de arrecadação para a União provenientes da mineração, com destaque para o mês de janeiro nos anos de 2024 e 2025, que registraram montantes superiores a 700 milhões de reais no Brasil. Em contrapartida, o mês de dezembro apresentou a menor arrecadação, possivelmente em razão da falta de consolidação de todos os dados referentes a esse período.

## Conclusão

A arrecadação da CFEM entre 2022 e 2025 mostra um fluxo de receitas elevado e relativamente estável, fortemente ancorado em minerais estratégicos como ferro, cobre e ouro, o que reforça a atratividade do setor mineral brasileiro para investidores em busca de proteção de capital e exposição a commodities em um cenário de instabilidade geopolítica. Ao mesmo tempo, a forte concentração da arrecadação em poucos estados e substâncias amplia a importância de análises de risco ligadas a ciclos de preços globais, conflitos internacionais e mudanças regulatórias que podem afetar margens, carga tributária efetiva e a previsibilidade de fluxos de caixa de projetos minerários. Nesse contexto, os dados da CFEM e a transparência oferecida pela ANM tornam-se insumos centrais para investidores avaliarem oportunidades de longo prazo, distinguindo ativos mais resilientes (por exemplo, em minerais críticos e com boa governança ambiental) de operações mais vulneráveis a choques de preço e pressões socioambientais.
